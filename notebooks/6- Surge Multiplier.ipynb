# 🚖 6.1- Smart Mobility Ecosystem: Final System Integration

### 📝 Project Overview
The final implementation represents a state-of-the-art **Smart Mobility Ecosystem**. It bridges the gap between raw Big Data and actionable, secure incentives. By integrating **Machine Learning logic** with **Blockchain technology**, the system ensures that driver performance is rewarded fairly, transparently, and autonomously.

---

### 🛠 Technical Pillars

#### 1. Feature Engineering & Dynamic Logic
The system goes beyond simple data processing by implementing intelligent "Surge" detection.
* **Temporal Analysis:** The `SurgeAI` module extracts the hour from the trip data to identify peak demand.
* **Dynamic Scaling:** A **1.5x Multiplier** is applied during rush hours (e.g., 18:00), ensuring the economic model adapts to real-time market needs.



#### 2. Advanced Software Architecture: Composition
We utilized the **Composition** design pattern to create a modular and scalable system.
* **Design Choice:** The `SmartTaxi` class "has-a" `SurgeAI` engine rather than just inheriting static traits.
* **Decoupling:** This allows the AI logic to be updated or replaced without altering the core vehicle or logistical classes, adhering to the **Open/Closed Principle**.



#### 3. Blockchain Security (The Immutable Ledger)
To solve the "Trust Gap" between drivers and operators, all rewards are finalized on a private Blockchain.
* **SHA-256 Hashing:** Every transaction is cryptographically linked to the previous block.
* **Tamper-Proof:** Once the reward for a driver (e.g., Anna) is recorded, it cannot be modified or deleted, providing 100% financial integrity.


```python
import hashlib
import json
import time

# 1. Blockchain Class (The Security)
class Blockchain:
    def __init__(self):
        self.chain = []
        self.balances = {}
        self.add_block(data="Genesis Block", prev_hash="0")

    def add_block(self, data, prev_hash=None):
        block = {
            'index': len(self.chain) + 1,
            'timestamp': time.time(),
            'data': data,
            'prev_hash': prev_hash or self.chain[-1]['hash']
        }
        block['hash'] = hashlib.sha256(json.dumps(block, sort_keys=True).encode()).hexdigest()
        self.chain.append(block)

    def add_reward(self, driver, amount, details):
        self.balances[driver] = self.balances.get(driver, 0) + amount
        self.add_block(data={"driver": driver, "reward": amount, "info": details})

# 2. SurgeAI Class (Feature Engineering Logic)
class SurgeAI: 
    def calculate_multiplier(self, hour):
        # Logic: 1.5x reward during peak hours
        if 8 <= hour <= 10 or 17 <= hour <= 20:
            return 1.5
        return 1.0

# 3. Main Taxi Classes (OOP & Composition)
class Taxi:
    def __init__(self, taxi_id):
        self.taxi_id = taxi_id

class SmartTaxi(Taxi):
    def __init__(self, taxi_id, driver_name):
        super().__init__(taxi_id) # Calling Taxi.__init__ with only taxi_id
        self.driver_name = driver_name
        self.surge_engine = SurgeAI() # Composition

    def process_smart_trip(self, hour, quality_score, blockchain):
        multiplier = self.surge_engine.calculate_multiplier(hour)
        base_reward = 10 if quality_score == "High" else 2
        final_reward = base_reward * multiplier
        
        blockchain.add_reward(
            driver=self.driver_name,
            amount=final_reward,
            details=f"Hour: {hour}, Multiplier: {multiplier}x"
        )
        return final_reward

# --- Execution (Test) ---
my_blockchain = Blockchain()
my_taxi = SmartTaxi("TX-99", "Anna")

# Testing a peak hour trip
reward = my_taxi.process_smart_trip(hour=18, quality_score="High", blockchain=my_blockchain)

print(f"✅ Success!")
print(f"Driver: {my_taxi.driver_name} | Taxi ID: {my_taxi.taxi_id}")
print(f"Final Reward at 18:00: {reward} Tokens")
print(f"Current Balance: {my_blockchain.balances['Anna']} Tokens")
```

# 🚖 6.2- Smart Mobility Ecosystem: Final Integrated Architecture

### 📝 Project Overview
This project represents a fully integrated **Autonomous Mobility Solution**. It successfully combines **Big Data processing**, **Machine Learning-driven incentives**, and **Blockchain-based security**. The system is designed to evaluate driver performance in real-time and secure rewards in an immutable ledger.

---

### 🛠 Technical Core Pillars

#### 1. Advanced Feature Engineering (AI Layer)
The system utilizes a `SmartIncentiveAI` module that performs two critical analytical tasks:
* **Efficiency Ratio:** Calculates `fare / distance` to determine trip quality. A ratio > 5.0 is classified as **"High Quality"**.
* **Temporal Analysis:** Implements **Dynamic Surge Logic**. During peak hours (8-10 AM and 5-8 PM), the system automatically applies a **1.5x multiplier** to rewards.



#### 2. Robust OOP Design (Architecture Layer)
The code follows professional software engineering patterns:
* **Composition:** The `Taxi` class "has-a" `SmartIncentiveAI` engine. This decoupling allows for independent updates to the AI logic without affecting the vehicle's core data.
* **Inheritance:** `ElectricTaxi` extends the base `Taxi` class, demonstrating a scalable hierarchy for diverse fleet types.



#### 3. Cryptographic Integrity (Security Layer)
To ensure absolute transparency between drivers and the company, we implemented a custom **Blockchain**:
* **SHA-256 Hashing:** Every trip record is hashed. Any attempt to modify a past reward would break the cryptographic chain.
* **Genesis to Reward:** The system initiates with a Genesis block and secures every reward (e.g., Anna's 15.0 Tokens) with a unique hash fingerprint.





```python
# --- Phase 1: Blockchain & Security Layer ---
class Blockchain:
    def __init__(self):
        self.chain = []
        self.balances = {}
        # Create Genesis Block
        self.add_block(previous_hash="0", data="Genesis Block")

    def add_block(self, data, previous_hash=None):
        block = {
            'index': len(self.chain) + 1,
            'timestamp': time.time(),
            'data': data,
            'previous_hash': previous_hash or self.chain[-1]['hash'] if self.chain else "0"
        }
        block['hash'] = self.generate_hash(block)
        self.chain.append(block)

    def generate_hash(self, block):
        block_string = json.dumps(block, sort_keys=True).encode()
        return hashlib.sha256(block_string).hexdigest()

    def add_reward(self, driver, amount, details):
        self.balances[driver] = self.balances.get(driver, 0) + amount
        data = {"driver": driver, "reward": amount, "info": details}
        self.add_block(data)

# --- Phase 2: AI & Feature Engineering Layer ---
class SmartIncentiveAI:
    def predict_quality(self, fare, distance):
        # Feature Engineering: Efficiency Ratio (Fare per Mile)
        efficiency = fare / distance if distance > 0 else 0
        return "High" if efficiency > 5.0 else "Normal"

    def get_surge_multiplier(self, hour):
        # Feature Engineering: Temporal Analysis (Peak Hours)
        if 8 <= hour <= 10 or 17 <= hour <= 20: # Rush hours
            return 1.5
        return 1.0

# --- Phase 3: OOP & Composition Layer ---
class Taxi:
    def __init__(self, taxi_id, driver_name):
        self.taxi_id = taxi_id
        self.driver_name = driver_name
        self.ai_engine = SmartIncentiveAI() # Composition: Taxi "has-a" AI Engine

class ElectricTaxi(Taxi): # Inheritance
    def process_trip(self, fare, distance, hour, blockchain):
        # 1. AI Quality Prediction
        quality = self.ai_engine.predict_quality(fare, distance)
        
        # 2. Dynamic Surge Multiplier (New Feature)
        multiplier = self.ai_engine.get_surge_multiplier(hour)
        
        # 3. Reward Calculation logic
        base_reward = 10 if quality == "High" else 2
        final_reward = base_reward * multiplier
        
        # 4. Immutable Blockchain Logging
        details = f"Quality: {quality}, Surge: {multiplier}x"
        blockchain.add_reward(self.driver_name, final_reward, details)
        
        return final_reward, quality, multiplier

# --- PHASE 4: EXECUTION & RESULTS ---
# Initialize System
taxi_chain = Blockchain()
my_taxi = ElectricTaxi("EV-2026", "Anna")

# Simulating a High-Quality Trip during Rush Hour (18:00)
fare, distance, hour = 35.50, 4.2, 18
reward, qual, surge = my_taxi.process_trip(fare, distance, hour, taxi_chain)

print(f"--- Trip Summary for {my_taxi.driver_name} ---")
print(f"Quality Prediction: {qual}")
print(f"Surge Multiplier: {surge}x")
print(f"Final Reward: {reward} Tokens")
print(f"Driver Balance: {taxi_chain.balances['Anna']} Tokens")
print(f"Last Block Hash: {taxi_chain.chain[-1]['hash']}")
```
